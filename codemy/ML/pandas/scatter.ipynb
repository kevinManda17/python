```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
# create a quick dataframe 
df = pd.DataFrame(randn(500, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
#scatter plot 
df.plot(kind="scatter", x="Mon", y="Tues")
```

```python
#scatter plot 
df.plot(kind="scatter", x="Mon", y="Tues", c="Wed")
```

```python
#Colors : https//matplotlib.org/stables/tutorials/colors/colormaps.html
df.plot(kind="scatter", x="Mon", y="Tues", c="Wed", cmap="magma")
```

```python
#By size 
df.plot(kind="scatter", x="Mon", y="Tues", s="Wed", cmap="magma")
```

```python
#By size 
df.plot(kind="scatter", x="Mon", y="Tues", s=df["Wed"]*100 , cmap="magma")
```

```python
#By size easier to read with alpha
df.plot(kind="scatter", x="Mon", y="Tues", s=df["Wed"]*100 , cmap="magma", alpha=0.4)
```

```python
#title
df.plot(kind="scatter", x="Mon", y="Tues", s=df["Wed"]*100 , cmap="magma", alpha=0.4, title="My Scatter Plot")
```
