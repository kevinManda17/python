```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
# create a quick dataframe 
df = pd.DataFrame(randn(25, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
df.abs().plot(kind = "bar")   
```

```python
#bar plot stacked 
df.abs().plot(kind = "bar" , stacked = True )   
```

```python
#shading
df.abs().plot(kind = "bar", stacked =True , alpha = 0.5)   
```

```python
#title
df.abs().plot(kind = "bar", stacked =True , alpha = 0.5 , title="My Bar Chart")   
```

```python
#legend 
df.abs().plot(kind = "bar", stacked =True , alpha = 0.5 , title="My Bar Chart", legend=False)   
```

```python
#second way 
df.abs().plot.bar()
```
