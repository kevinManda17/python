```python
import pandas as pd 
import numpy as np 
#Crete Data 
data= {
    'Corporation': ['Apple','Google','IBM', 'Apple','Google','IBM'],
    'Employees': ['Kevin', 'Keven', 'Iron Manda', 'Mubenga','kev', "Kevin Manda"],
    'Salary': [100000, 200000, 300000, 400000, 500000, 600000],
}
#create DateFrame 
df= pd.DataFrame(data)
df 
```

```python
#create function 
def times1000(x):
    return format(x*1000, ",d") 
```

```python
#Use Fucntion 
df['Salary'].apply(times1000)
```

```python
#Make it into the Dataframe 
pd.DataFrame(df['Salary'].apply(times1000))
```

```python
#Add Manda Last Name 
def namer (x):
    if x == "Kevin" :
        return "IRON MANDA 3.0"
    else:  
        return x
```

```python
#shoe manda last name 
df["Employees"].apply(namer)
```

```python
#use lambda
pd.DataFrame(df['Salary'].apply(lambda x : format(x*12000, ",d")))
```
