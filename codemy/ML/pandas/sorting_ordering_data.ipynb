```python
import pandas as pd 
import numpy as np 
#Crete Data 
data= {
    'Corporation': ['Apple','Google','IBM', 'Apple','Google','IBM'],
    'Employees': ['Kevin', 'Keven', 'Iron Manda', 'Mubenga','kev', "Kevin Manda"],
    'Salary': [100, 200, 300, 400, 500, 600],
}
#create DateFrame 
df= pd.DataFrame(data)
df 
```

```python
#sort salaries from lowest to hightest 
df.sort_values('Salary')
```

```python
df.sort_values('Salary', ascending=False)
```

```python
df.sort_values('Salary', ascending=False, inplace=True )
df
```

```python
#sort strings Alphabetically
df.sort_values("Corporation")
```

```python
#sort strings Alphabetically Descending 
df.sort_values("Corporation", ascending=False)
```
