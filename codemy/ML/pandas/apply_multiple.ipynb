```python
import pandas as pd 
import numpy as np 
#Crete Data 
data= {
    'Corporation': ['Apple','Google','IBM', 'Apple','Google','IBM'],
    'Employees': ['Kevin', 'Keven', 'Iron Manda', 'Mubenga','kev', "Kevin Manda"],
    'Salary': [100000, 200000, 300000, 400000, 500000, 600000],
}
#create DateFrame 
df= pd.DataFrame(data)
df 
```

```python
#create function 
def times10(x):
    return format(x*1000, ",d") 
```

```python
#Use apply Functions 
df['Salary'].apply(times10)
```

```python
#append to current DF 
df['Salary'] = df['Salary'].apply(times10)
df
```

```python
def namer(x ):
    return "DFL: " + x
```

```python
df[['Corporation', 'Employees']].apply(namer)
```

```python
#append to current DF 
df[['Corporation', 'Employees']] = df[['Corporation', 'Employees']].apply(namer)
df
```
