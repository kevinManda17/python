```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
# create a quick dataframe 
df = pd.DataFrame(randn(500, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
#boxplots 
df.plot(kind="box")
```

```python
#boxplot size 
df.plot(kind="box", figsize=(10, 5))
```

```python
#boxplot color 
df.plot(kind="box", figsize=(10, 5), cmap="jet")
```

```python
#No Legend 
df.plot(kind="box", legend=False)
```

```python
#title 
df.plot(kind="box", title="Boxplot")
```

```python
df.plot.box(title="my Box Plot")
```
