```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
# create a quick dataframe 
df = pd.DataFrame(randn(500, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
#Hexplots
df.plot(kind="hexbin", x="Mon", y="Tues")
```

```python
#Hexplots gridsize 
df.plot(kind="hexbin", x="Mon", y="Tues", gridsize=25)
```

```python
#color 
df.plot(kind="hexbin", x="Mon", y="Tues", gridsize=25 , cmap="Wistia")
```

```python
#othr method 
df.plot.hexbin(x="Mon", y="Tues", gridsize=10)
```

```python
#
df.plot.hexbin(x="Mon", y="Tues", gridsize=10, title="My Hex Plot")
```
