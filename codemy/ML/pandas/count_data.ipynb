```python
import pandas as pd 
import numpy as np 
```

```python
# import csv file 
my_df = pd.read_csv('dog_data_2670.csv')
my_df
```

```python
# count distincts values -descending 
my_df['Color'].value_counts()
```

```python
# count distincts values ascending 
my_df["Color"].value_counts(ascending=True)
```

```python
# count distincts values -descending 
my_df['Color'].value_counts(dropna=False)
```

```python
# get relative frequency 
my_df['Color'].value_counts(normalize=True)
```

```python
# get specific Itme count 
my_df['Color'].value_counts()['White']
```

```python
# Count Unique Values - size 
my_df.groupby('Color').size()
```

```python
# count unique values - count
my_df.groupby('Color').count()
```
