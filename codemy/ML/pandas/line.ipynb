```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
# create a quick dataframe 
df = pd.DataFrame(randn(25, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
#line
df.plot.line() 
```

```python
#line plot absolute values
df.abs().plot.line() 
```

```python
#line width 
df.abs().plot(kind="line", linewidth=2)
```

```python
df['Mon'].abs().plot(kind="line", linewidth=2)
```

```python
df[['Mon', 'Tues']].abs().plot(kind="line", linewidth=2)
```

```python
df[['Mon', 'Tues']].abs().plot(kind="line", lw=2, figsize=(10, 5))
```

```python
#title and stacked 
df[['Mon', 'Tues']].abs().plot(kind="line", lw=2, figsize=(10, 5), title="My line plot")
```

```python
#second way line 
df.plot.line()
```
