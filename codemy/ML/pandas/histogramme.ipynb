```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
#create  a quick dataframe 
df = pd.DataFrame(randn(100, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
#histogram 
df["Wed"].hist()
```

```python
#histgram with smaller bins , 50 , 20 
df["Wed"].hist(bins=20)
```

```python
#another way without grid bars 
df["Wed"].plot(kind="hist", bins=20 , legend =True)
```
