```python
import pandas as pd 
import numpy as np 
#create Dummy Data
stuff = {'A':[1,2,3], 'B':[4,np.nan, 6], 'C':[7,8,9], 'D':[10,11,12]} 
df = pd.DataFrame(stuff)
df
```

```python
#drop Rows with null data 
df.dropna()
```

```python
#drop columns with null data 
df.dropna(axis=1)
```

```python
#not permanent , inplce =False 
df
```

```python
#more than one ? set Threshold
df.dropna(thresh=2, axis=1)
```

```python
#Replace thin with fillna()
df.fillna(value="IRON MANDA")
```

```python
#use math functions 
df['B'].fillna(value=df['B'].mean())
```

```python
#use min max 
df['B'].fillna(value=df['B'].max())
```

```python
#use min 
df['B'].fillna(value=df['B'].min())
```

```python
#use sum 
df['B'].fillna(value=df['B'].sum())
```
