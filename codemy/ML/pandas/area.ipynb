```python
import pandas as pd 
import numpy as np 
from numpy.random import randn
%matplotlib inline 
# create a quick dataframe 
df = pd.DataFrame(randn(25, 4 ), columns = ["Mon", "Tues", "Wed", "Thur"])
df
```

```python
#area plot
df.plot(kind="area",stacked=False)  
```

```python
#area plot 
df.abs().plot(kind="area")  
```

```python
#shading 
df.abs().plot(kind="area", alpha=0.5)  
```

```python
#title 
df.abs().plot(kind="area", alpha=0.5, title="My Area Plot")  
```

```python
#legend 
df.abs().plot(kind="area", alpha=0.5, title="My Area Plot", legend=False)  
```

```python
df["Mon"].abs().plot(kind="area", alpha=0.5, title="My Area Plot")  
```

```python
#plot 2 columns 
df[["Mon", "Thur"]].abs().plot(kind="area", alpha=0.5, title="My Area Plot")  
```

```python
df.abs().plot(kind="area", alpha=0.5, title="My Area Plot", table = True)  
```
